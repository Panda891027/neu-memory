# 🚀 neu-memory Quick Start

Welcome! This interactive notebook will guide you through using neu-memory.

## What is neu-memory?

neu-memory provides long-term memory for LLM agents. It allows AI assistants to:
- 📝 Create and manage persistent files
- 🔍 Read and search through saved information
- ✏️ Edit content with precision
- 🗂️ Organize memories in directories

All under the `/memories` path!

## Step 1: Setup

First, let's import the necessary modules and create a memory tool.

```python
import sys
from pathlib import Path

# Add the package to path if running in Codespaces
sys.path.insert(0, str(Path.cwd()))

from neu_memory import create_memory_tool
from neu_memory_storage_in_memory import InMemoryStorage

# Initialize storage and memory tool
storage = InMemoryStorage()
memory_kit = create_memory_tool(storage)

print("✅ Memory tool initialized!")
print(f"Tool name: {memory_kit.name}")
```

## Step 2: Create Your First Memory

Let's create a simple note file.

```python
result = await memory_kit.execute({
    "command": "create",
    "path": "/memories/welcome.txt",
    "file_text": "Hello from neu-memory!\nThis is my first memory.",
})

print(result)
```

## Step 3: View Your Memory

Now let's read the file we just created.

```python
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories/welcome.txt",
})

print(result)
```

## Step 4: Edit with String Replace

Let's update part of our note using `str_replace`.

```python
result = await memory_kit.execute({
    "command": "str_replace",
    "path": "/memories/welcome.txt",
    "old_str": "first memory",
    "new_str": "first memory, and it's amazing!",
})

print(result)

# View the updated content
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories/welcome.txt",
})

print("\nUpdated content:")
print(result)
```

## Step 5: Insert Text at Specific Lines

You can insert text at any line number.

```python
result = await memory_kit.execute({
    "command": "insert",
    "path": "/memories/welcome.txt",
    "insert_line": 2,
    "insert_text": "Created on: 2025-01-15",
})

print(result)

# View the result
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories/welcome.txt",
})

print("\nUpdated content:")
print(result)
```

## Step 6: Create Organized Structure

Let's create a more organized memory structure with subdirectories.

```python
# Create a project log
await memory_kit.execute({
    "command": "create",
    "path": "/memories/projects/my-app/log.txt",
    "file_text": "Project Log\n==========\n- Started development\n- Added core features",
})

# Create a to-do list
await memory_kit.execute({
    "command": "create",
    "path": "/memories/projects/my-app/todo.txt",
    "file_text": "TODO:\n[ ] Add tests\n[ ] Write documentation\n[ ] Deploy to production",
})

print("✅ Created project structure!")
```

## Step 7: Browse Directory Contents

View what's in a directory.

```python
# View root directory
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories",
})

print("Root directory:")
print(result)

# View project directory
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories/projects/my-app",
})

print("\nProject directory:")
print(result)
```

## Step 8: Rename/Move Files

Reorganize your memories by renaming or moving files.

```python
result = await memory_kit.execute({
    "command": "rename",
    "old_path": "/memories/welcome.txt",
    "new_path": "/memories/intro.txt",
})

print(result)

# Verify the change
result = await memory_kit.execute({
    "command": "view",
    "path": "/memories",
})

print("\nRoot directory after rename:")
print(result)
```

## Step 9: Delete When Done

Clean up files you no longer need.

```python
result = await memory_kit.execute({
    "command": "delete",
    "path": "/memories/intro.txt",
})

print(result)
```
